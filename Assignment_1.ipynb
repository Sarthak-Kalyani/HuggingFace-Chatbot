```python
# Task 1 — Hugging Face Integration

# Import Required Libraries
import torch
from transformers import (AutoTokenizer, AutoModelForCausalLM, pipeline)
```

```python
# Load Instruction-Tuned Model
model_name = "Qwen/Qwen2.5-1.5B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name, torch_dtype="auto", device_map="auto")
```

```python
# Create Text Generation Pipeline
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer)
```

```python
# Test the Model
prompt = "Patient has fever for 3 days."
response = pipe(prompt,max_new_tokens=150,do_sample=True, temperature=0.7)
print(response[0]["generated_text"])
```

```python
# AI Healthcare Chatbot
print("===== AI Healthcare Assistant =====")
print("Type 'exit' to stop.\n")
while True:
    question = input("You: ")
    if question.lower() == "exit":
        print("Assistant: Thank you! Stay healthy.")
        break
    prompt = f"You are an AI Healthcare Assistant.\nQuestion:
    {question}\nAnswer:"
    answer = pipe(prompt, max_new_tokens=200, temperature=0.7)
    print("\nAssistant:")
    print(answer[0]["generated_text"])
    print("-" * 50)
```

```python
# Task 2 : Prompt Engineering
prompt = """
You are an experienced medical AI assistant.

Context:
A medical student wants to learn about diabetes.

Instructions:
- Give only educational information.
- Do not diagnose the patient.
- Keep the answer simple.
- Format the answer in Markdown.

Include these headings:
# Disease Overview
# Symptoms
# Risk Factors
# Prevention
# When to Consult a Doctor
"""
response = pipe(prompt, max_new_tokens=400)
print(response[0]["generated_text"])
```

```python
# Task 3 : Few-Shot Prompting

prompt = """
Convert the clinical notes into the same format as the examples.

Example 1
Input: Patient has cough and fever.

Output:
Symptoms:
- Cough
- Fever
Severity: Moderate

Example 2
Input: Patient has chest pain and shortness of breath.

Output:
Symptoms:
- Chest pain
- Shortness of breath
Severity: High

Example 3
Input: Patient has headache and mild dizziness.

Output:
Symptoms:
- Headache
- Mild dizziness
Severity: Low

Now convert this.

Input: Patient has fever, sore throat and body pain.

Output:
"""
response = pipe(prompt,max_new_tokens=100, do_sample=False,
                return_full_text=False)
print(response[0]["generated_text"])
```

```python
# Task 4 : Chain of Thought Prompting

prompt = """
You are an experienced AI Healthcare Assistant.
A patient has:
- Fever
- Cough
- Oxygen Level = 88%

Instructions:
1. Carefully analyze the patient's symptoms.
2. Consider possible medical conditions.
3. Identify any warning signs or risk factors.
4. Based on your analysis, provide the most appropriate recommendation.
5. Do not reveal your internal reasoning process.
6. Keep the response clear, concise, and suitable for a healthcare professional.

Provide the final recommendation in Markdown format.
"""
print(prompt)
```

## Observation

This prompt encourages the AI model to analyze the patient's condition before providing a recommendation. The prompt explicitly instructs the model not to reveal its internal reasoning while still producing a clear and structured final response.

```python
# Task 5 : Structured Output
prompt = """
You are an AI Healthcare Assistant.
Analyze the patient's medical condition and return the response ONLY in the following JSON format.
{
    "disease": "",
    "symptoms": [],
    "risk_level": "",
    "recommendation": ""
}
Do not include any explanation, headings, or extra text.
Return only valid JSON.
"""
print(prompt)
```

## Observation

This prompt instructs the AI model to generate the response in a fixed JSON format. Using structured output makes the response easier to process, store, and integrate into healthcare applications and APIs.

```python
# Task 7 : Prompt Optimization
poor_response = """
Maybe the patient has malaria.
Or maybe dengue.
Or maybe COVID.
I am not sure.
"""
print("Poor Response:")
print(poor_response)

print("\nWhy is the response poor?")
print("- It gives multiple possible diseases without proper analysis.")
print("- It is vague and lacks confidence.")
print("- It provides no explanation or recommendation.")
print("- It may confuse the user and can lead to hallucinations.")

print("\nPrompt Engineering Techniques:")
print("- Role Prompting")
print("- Context")
print("- Constraints")
print("- Structured Output")

optimized_prompt = """
You are an experienced AI Healthcare Assistant.
Analyze the patient's symptoms carefully.

Instructions:
- Use only the information provided.
- Do not guess or invent diseases.
- If there is insufficient information, clearly say that more clinical details are required.
- Explain the possible condition briefly.
- Recommend consulting a healthcare professional.
- Keep the response clear and concise.
"""
print("\nOptimized Prompt:\n")
print(optimized_prompt)
```

## Observation

The original response is unclear because it lists several diseases without evidence or recommendations. The improved prompt uses role prompting, context, constraints, and clear instructions to reduce hallucinations and generate a more reliable response.

```python
# Bonus Challenge : Role-Based Prompt Template
prompt_template = """
You are an AI Healthcare Assistant.
User Role: {role}
Question:
{question}

Instructions:
- If the role is Doctor:
  Provide a detailed clinical explanation using medical terminology.

- If the role is Nurse:
  Focus on patient care, monitoring, and nursing responsibilities.

- If the role is Medical Student:
  Explain the concept in a simple educational manner with brief examples.

- If the role is Patient:
  Use simple, easy-to-understand language and avoid complex medical terms.

Keep the response accurate, clear, and helpful.
"""
print(prompt_template)
```

## Observation
The same Hugging Face model can provide different responses by changing only the prompt. The user's role is passed as a variable, allowing the chatbot to adapt its language and level of detail without changing the underlying model.
